```python
from pathlib import Path
import shutil
import json
import torch

from huggingface_hub import HfApi

# Persistent storage configuration


HF_REPO_ID = "shashik-bits/neuroBrain-checkpoints"

LOCAL_CHECKPOINT_DIR = Path(
    "/content/neuroBrain/results/checkpoints/formal_experiment"
)
LOCAL_CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

HF_BACKUP_DIR = "formal_experiment"

api = HfApi()

# Helper: upload a checkpoint safely

def upload_checkpoint(local_path, repo_filename):
    local_path = Path(local_path)

    api.upload_file(
        path_or_fileobj=str(local_path),
        path_in_repo=f"{HF_BACKUP_DIR}/{repo_filename}",
        repo_id=HF_REPO_ID,
        repo_type="model",
    )

    print(f"Uploaded: {repo_filename}")



# Helper: save a complete recoverable checkpoint


def save_persistent_checkpoint(
    epoch,
    train_loss,
    val_results,
    best_metric,
):
    checkpoint = {
        "epoch": epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": scheduler.state_dict(),
        "train_loss": train_loss,
        "val_results": val_results,
        "best_metric": best_metric,
        "learning_rate": optimizer.param_groups[0]["lr"],
    }

    latest_path = LOCAL_CHECKPOINT_DIR / "latest.pt"

    torch.save(checkpoint, latest_path)

    upload_checkpoint(
        latest_path,
        "latest.pt",
    )

    # Save a separate epoch checkpoint.
    epoch_path = LOCAL_CHECKPOINT_DIR / f"epoch_{epoch:02d}.pt"

    torch.save(checkpoint, epoch_path)

    upload_checkpoint(
        epoch_path,
        f"epoch_{epoch:02d}.pt",
    )

    # Save best checkpoint.
    if val_results["mean_tumor_dice"] >= best_metric:
        best_path = LOCAL_CHECKPOINT_DIR / "best_model.pt"

        torch.save(checkpoint, best_path)

        upload_checkpoint(
            best_path,
            "best_model.pt",
        )

        return val_results["mean_tumor_dice"]

    return best_metric


print("Persistent checkpoint system initialized.")
print("Local:", LOCAL_CHECKPOINT_DIR)
print("Hugging Face:", HF_REPO_ID)
```

```python
from pathlib import Path
import json
import torch
from huggingface_hub import HfApi

HF_REPO_ID = "shashik-bits/neuroBrain-checkpoints"

LOCAL_CHECKPOINT_DIR = Path(
    "/content/neuroBrain/results/checkpoints/formal_experiment"
)
LOCAL_CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

HF_BACKUP_DIR = "formal_experiment"

api = HfApi()


def upload_checkpoint(local_path, repo_filename):
    local_path = Path(local_path)

    api.upload_file(
        path_or_fileobj=str(local_path),
        path_in_repo=f"{HF_BACKUP_DIR}/{repo_filename}",
        repo_id=HF_REPO_ID,
        repo_type="model",
    )

    print(f"Uploaded: {repo_filename}")


def save_persistent_checkpoint(
    epoch,
    train_loss,
    val_results,
    best_metric,
):
    checkpoint = {
        "epoch": epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": scheduler.state_dict(),
        "train_loss": train_loss,
        "val_results": val_results,
        "best_metric": best_metric,
        "learning_rate": optimizer.param_groups[0]["lr"],
    }


    # 1. Latest checkpoint — uploaded EVERY epoch


    latest_path = LOCAL_CHECKPOINT_DIR / "latest.pt"

    torch.save(checkpoint, latest_path)

    upload_checkpoint(
        latest_path,
        "latest.pt",
    )

    # 2. Training history — uploaded EVERY epoch


    history_path = LOCAL_CHECKPOINT_DIR / "training_history.json"

    with open(history_path, "w") as f:
        json.dump(trainer.history, f, indent=2)

    upload_checkpoint(
        history_path,
        "training_history.json",
    )


    # 3. Best checkpoint — ONLY when improved

    current_metric = val_results["mean_tumor_dice"]

    if current_metric > best_metric:

        best_path = LOCAL_CHECKPOINT_DIR / "best_model.pt"

        torch.save(checkpoint, best_path)

        upload_checkpoint(
            best_path,
            "best_model.pt",
        )

        print(
            f"   New best mean tumor Dice: "
            f"{current_metric:.4f}"
        )

        return current_metric

    return best_metric


print("Updated persistent checkpoint system ready.")
print("Uploads per epoch:")
print("  latest.pt")
print("  training_history.json")
print("Best model:")
print("  best_model.pt only when improved")
```

```python
from pathlib import Path
import json
import torch
from huggingface_hub import HfApi

# Persistence configuration


# HF_REPO_ID = "shashik-bits/neuroBrain-checkpoints"

LOCAL_CHECKPOINT_DIR = Path(
    "/content/neuroBrain/results/checkpoints/formal_experiment"
)
LOCAL_CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

HF_BACKUP_DIR = "formal_experiment"

api = HfApi()


def upload_checkpoint(local_path, repo_filename):
    local_path = Path(local_path)

    api.upload_file(
        path_or_fileobj=str(local_path),
        path_in_repo=f"{HF_BACKUP_DIR}/{repo_filename}",
        repo_id=HF_REPO_ID,
        repo_type="model",
    )

    print(f"Uploaded: {repo_filename}")


def save_persistent_checkpoint(
    epoch,
    train_loss,
    val_results,
    best_metric,
):

    checkpoint = {
        "epoch": epoch,
        "model_state_dict": model.state_dict(),
        "optimizer_state_dict": optimizer.state_dict(),
        "scheduler_state_dict": scheduler.state_dict(),
        "train_loss": train_loss,
        "val_results": val_results,
        "best_metric": best_metric,
        "learning_rate": optimizer.param_groups[0]["lr"],
    }


    # Save latest checkpoint


    latest_path = LOCAL_CHECKPOINT_DIR / "latest.pt"

    torch.save(checkpoint, latest_path)

    upload_checkpoint(
        latest_path,
        "latest.pt",
    )

    # Save training history


    history_path = LOCAL_CHECKPOINT_DIR / "training_history.json"

    with open(history_path, "w") as f:
        json.dump(trainer.history, f, indent=2)

    upload_checkpoint(
        history_path,
        "training_history.json",
    )


    # Save best checkpoint if improved


    current_metric = val_results["mean_tumor_dice"]

    if current_metric > best_metric:

        best_path = LOCAL_CHECKPOINT_DIR / "best_model.pt"

        torch.save(checkpoint, best_path)

        upload_checkpoint(
            best_path,
            "best_model.pt",
        )

        print(
            f"New best mean tumor Dice: "
            f"{current_metric:.4f}"
        )

        return current_metric

    return best_metric


print("Persistent checkpoint system restored successfully.")
```
